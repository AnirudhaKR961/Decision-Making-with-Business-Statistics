# Problem 2

The breaking strength of gunny bags used for packaging cement is normally distributed with a mean of 5 kg per sq. centimeter and a standard deviation of 1.5 kg per sq. centimeter. The quality team of the cement company wants to know the following about the packaging material to better understand wastage or pilferage within the supply chain.

## 2.1 What proportion of the gunny bags have a breaking strength of less than 3.17 kg per sq cm?

```python
# Given values
mean = 5
std_dev = 1.5

# Calculate probability
probability = stats.norm.cdf(3.17, loc=mean, scale=std_dev)

print("Proportion of bags with strength < 3.17:", probability)
```

Probability = 0.111 (≈ 11.1%)

```python
# Given values
mean = 5
std_dev = 1.5

# X range
x = np.linspace(0, 10, 1000)
y = stats.norm.pdf(x, mean, std_dev)

# Cutoff value
cutoff = 3.17
```

```python
# Plot normal curve
plt.plot(x, y)

# Shade LEFT region (P(X < 3.17))
x_shade = np.linspace(0, cutoff, 500)
y_shade = stats.norm.pdf(x_shade, mean, std_dev)
plt.fill_between(x_shade, y_shade, color='red')

# Vertical line at cutoff
plt.axvline(x=cutoff, c="black")

# Text labels
plt.text(0.5, 0.1, "Gunny bags\nless than 3.17")
plt.text(4, 0.1, "Gunny bags\nmore than 3.17")

# Labels
plt.title("Normal Distribution (Left-Tailed)")
plt.xlabel("Breaking Strength")
plt.ylabel("Density")

plt.show()
```

## 2.2 What proportion of the gunny bags have a breaking strength of at least 3.6 kg per sq cm.?

```python
# Given values
mean = 5
std_dev = 1.5
cutoff = 3.6

# Probability
probability = 1 - stats.norm.cdf(cutoff, mean, std_dev)
print("P(X >= 3.6):", probability)
```

Probability = 0.824 (≈ 82.4%)

```python
# Plot
x = np.linspace(0, 10, 1000)
y = stats.norm.pdf(x, mean, std_dev)

plt.figure()
plt.plot(x, y)

# Shade RIGHT region
x_shade = np.linspace(cutoff, 10, 500)
y_shade = stats.norm.pdf(x_shade, mean, std_dev)
plt.fill_between(x_shade, y_shade, color='pink')

# Vertical line
plt.axvline(x=cutoff, c='purple')

# Labels
plt.title("Normal Distribution (Right-Tailed: X ≥ 3.6)")
plt.xlabel("Breaking Strength")
plt.ylabel("Density")

plt.show()
```

## 2.3 What proportion of the gunny bags have a breaking strength between 5 and 5.5 kg per sq cm.?

```python
# Given values
mean = 5
std_dev = 1.5

# Range
lower = 5
upper = 5.5

# Probability
probability = stats.norm.cdf(upper, mean, std_dev) - stats.norm.cdf(lower, mean, std_dev)
print("P(5 \u2264 X \u2264 5.5):", probability)
```

Probability = 0.130 (≈ 13.0%)

```python
# Plot
x = np.linspace(0, 10, 1000)
y = stats.norm.pdf(x, mean, std_dev)

plt.figure()
plt.plot(x, y)

# Shade region between 5 and 5.5
x_shade = np.linspace(lower, upper, 500)
y_shade = stats.norm.pdf(x_shade, mean, std_dev)
plt.fill_between(x_shade, y_shade, color='grey')

# Vertical lines
plt.axvline(x=lower, c='black')
plt.axvline(x=upper, c='black')

# Labels
plt.title("Normal Distribution (5 \u2264 X \u2264 5.5)")
plt.xlabel("Breaking Strength")
plt.ylabel("Density")

plt.show()
```

## 2.4 What proportion of the gunny bags have a breaking strength NOT between 3 and 7.5 kg per sq cm.?

```python
# Given values
mean = 5
std_dev = 1.5

# Range
lower = 3
upper = 7.5

# Probability (outside range)
probability = stats.norm.cdf(lower, mean, std_dev) + (1 - stats.norm.cdf(upper, mean, std_dev))
print("P(X < 3 or X > 7.5):", probability)
```

Probability = 0.139 (≈ 13.9%)

```python
# Plot
x = np.linspace(0, 10, 1000)
y = stats.norm.pdf(x, mean, std_dev)

plt.figure()
plt.plot(x, y)

# Shade LEFT tail
x_left = np.linspace(0, lower, 500)
y_left = stats.norm.pdf(x_left, mean, std_dev)
plt.fill_between(x_left, y_left, color='violet')

# Shade RIGHT tail
x_right = np.linspace(upper, 10, 500)
y_right = stats.norm.pdf(x_right, mean, std_dev)
plt.fill_between(x_right, y_right, color='purple')

# Vertical lines
plt.axvline(x=lower, c='black')
plt.axvline(x=upper, c='black')

# Labels
plt.title("Normal Distribution (Outside 3 to 7.5)")
plt.xlabel("Breaking Strength")
plt.ylabel("Density")

plt.show()
```

# Problem 4

Dental implant data: The hardness of metal implants in dental cavities depends on multiple factors, such as the method of implant, the temperature at which the metal is treated, the alloy used as well as the dentists who may favor one method above another and may work better in his/her favorite method. ## 4.1 How does the hardness of implants vary depending on dentists?



```python
from scipy.stats import shapiro, levene, f_oneway
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Preview data
print(data.head())

# Function to analyze each alloy
def analyze_alloy(data_df, alloy_id):
    print(f"\n===== ANALYSIS FOR ALLOY {alloy_id} =====\n")

    subset = data_df[data_df['Alloy'] == alloy_id]

    # Assumption 1: Normality
    print("Normality Test (Shapiro-Wilk):")
    for dentist in subset['Dentist'].unique():
        stat, p = shapiro(subset[subset['Dentist'] == dentist]['Response'])
        print(f"Dentist {dentist}: p = {p:.4f}")

    # Assumption 2: Homogeneity of Variances
    groups = [group['Response'].values for name, group in subset.groupby('Dentist')]
    stat, p = levene(*groups)
    print(f"\nLevene’s Test (Equal Variance): p = {p:.4f}")

    # ANOVA Test
    stat, p = f_oneway(*groups)
    print(f"\nANOVA Result: p-value = {p:.4f}")

    if p < 0.05:
        print("Reject H0: Hardness differs among dentists")

        # Post-hoc Test (Tukey)
        tukey = pairwise_tukeyhsd(
            endog=subset['Response'],
            groups=subset['Dentist'],
            alpha=0.05
        )
        print("\nTukey Test Results:")
        print(tukey)

    else:
        print("Fail to reject H0: No significant difference")

    # Boxplot Visualization
    plt.figure(figsize=(10, 5))
    sns.boxplot(x='Dentist', y='Response', data=subset)
    sns.stripplot(x='Dentist', y='Response', data=subset, color='black', alpha=0.5)
    plt.title(f'Hardness Distribution by Dentist (Alloy {alloy_id})')
    plt.ylabel('Hardness Response')
    plt.show()

    # Mean plot
    plt.figure(figsize=(10, 5))
    mean_hardness = subset.groupby('Dentist')['Response'].mean().reset_index()
    sns.barplot(x='Dentist', y='Response', data=mean_hardness, palette='viridis')
    plt.title(f'Mean Hardness by Dentist (Alloy {alloy_id})')
    plt.ylabel('Mean Hardness Response')
    plt.show()

# Get unique alloys and run analysis for each
for alloy in data['Alloy'].unique():
    analyze_alloy(data, alloy)
```

## 4.2 How does the hardness of implants vary depending on methods?


```python
from scipy.stats import shapiro, levene, f_oneway
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# ==============================
# LOAD DATA
# ==============================
file_path = '/content/drive/MyDrive/Assessments   projects/Second Project (Decision Making with Business Statistics)/Dental Hardness data.xlsx'
df = pd.read_excel(file_path)

# Rename for clarity
df = df.rename(columns={"Response": "Hardness"})

# ==============================
# FUNCTION FOR ANALYSIS
# ==============================
def analyze_method(data, alloy):

    print("\n" + "="*50)
    print(f"ANALYSIS FOR ALLOY {alloy}")
    print("="*50)

    subset = data[data["Alloy"] == alloy]

    # --------------------------
    # ASSUMPTION CHECK
    # --------------------------
    print("\nNormality Test (Shapiro):")
    for m in subset["Method"].unique():
        stat, p = shapiro(subset[subset["Method"] == m]["Hardness"])
        print(f"Method {m}: p = {p:.4f}")

    groups = [g["Hardness"].values for _, g in subset.groupby("Method")]
    stat, p = levene(*groups)
    print(f"\nLevene Test (Equal Variance): p = {p:.4f}")

    # --------------------------
    # ANOVA TEST
    # --------------------------
    f_stat, p_val = f_oneway(*groups)
    print(f"\nANOVA p-value = {p_val:.6f}")

    if p_val < 0.05:
        print("Reject H0 (Significant differences exist)")

        tukey = pairwise_tukeyhsd(
            endog=subset["Hardness"],
            groups=subset["Method"],
            alpha=0.05
        )
        print("\nTukey Test:")
        print(tukey)

    else:
        print("Fail to reject H0 (No significant difference)")

    # ==============================
    # CREATIVE GRAPHS
    # ==============================

    # 1Scatter Plot (Distribution)
    plt.figure()
    for m, g in subset.groupby("Method"):
        plt.scatter([m]*len(g), g["Hardness"])
    plt.title(f"Alloy {alloy} - Hardness Distribution by Method")
    plt.xlabel("Method")
    plt.ylabel("Hardness")
    plt.show()

    # Boxplot (Spread + Median)
    plt.figure()
    subset.boxplot(column="Hardness", by="Method")
    plt.title(f"Alloy {alloy} - Boxplot")
    plt.suptitle("")
    plt.xlabel("Method")
    plt.ylabel("Hardness")
    plt.show()

    # Mean Plot with Error Bars
    means = subset.groupby("Method")["Hardness"].mean()
    std = subset.groupby("Method")["Hardness"].std()

    plt.figure()
    plt.errorbar(means.index, means.values, yerr=std.values, fmt='o')
    plt.title(f"Alloy {alloy} - Mean Hardness with Variability")
    plt.xlabel("Method")
    plt.ylabel("Mean Hardness")
    plt.show()

    # Line Trend Plot
    plt.figure()
    plt.plot(means.index, means.values, marker='o')
    plt.title(f"Alloy {alloy} - Trend of Mean Hardness")
    plt.xlabel("Method")
    plt.ylabel("Mean Hardness")
    plt.show()


# ==============================
# RUN FOR BOTH ALLOYS
# ==============================
for alloy in df["Alloy"].unique():
    analyze_method(df, alloy)
```

## 4.3 What is the interaction effect between the dentist and method on the hardness of dental implants for each type of alloy?


```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import statsmodels.api as sm
from statsmodels.formula.api import ols

# Load dataset
file_path = '/content/drive/MyDrive/Assessments   projects/Second Project (Decision Making with Business Statistics)/Dental Hardness data.xlsx'   # update path if needed
df = pd.read_excel(file_path)

# Rename 'Response' column to 'Hardness' for consistency
df = df.rename(columns={"Response": "Hardness"})

# Preview data
print(df.head())

# Ensure categorical variables
df['Dentist'] = df['Dentist'].astype('category')
df['Method'] = df['Method'].astype('category')
df['Alloy'] = df['Alloy'].astype('category')

# Get unique alloys
alloys = df['Alloy'].unique()

# Loop through each alloy separately
for alloy in alloys:
    print("\n" + "="*50)
    print(f"Analysis for Alloy: {alloy}")
    print("="*50)

    # Filter data for one alloy
    data = df[df['Alloy'] == alloy]

    # ----------------------------
    # 1. Interaction Plot
    # ----------------------------
    plt.figure(figsize=(8,6))
    sns.pointplot(data=data, x='Method', y='Hardness',
                  hue='Dentist', dodge=True, markers='o', capsize=0.1)

    plt.title(f'Interaction Plot: Dentist vs Method (Alloy {alloy})')
    plt.xlabel('Method')
    plt.ylabel('Mean Hardness')
    plt.legend(title='Dentist')
    plt.grid(True)
    plt.show()

    # ----------------------------
    # 2. Two-Way ANOVA with Interaction
    # ----------------------------
    model = ols('Hardness ~ C(Dentist) * C(Method)', data=data).fit()
    anova_table = sm.stats.anova_lm(model, typ=2)

    print("\nANOVA Table:")
    print(anova_table)
```

## 4.4 How does the hardness of implants vary depending on dentists and methods together?


```python
# Import libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from statsmodels.formula.api import ols
import statsmodels.api as sm
from scipy.stats import shapiro, levene
from statsmodels.stats.multicomp import pairwise_tukeyhsd

# Load dataset
df = pd.read_excel('/content/drive/MyDrive/Assessments   projects/Second Project (Decision Making with Business Statistics)/Dental Hardness data.xlsx')

# Rename 'Response' column to 'Hardness' for consistency
df = df.rename(columns={"Response": "Hardness"})

# Convert to categorical
df['Dentist'] = df['Dentist'].astype('category')
df['Method'] = df['Method'].astype('category')
df['Alloy'] = df['Alloy'].astype('category')

# Loop for each alloy
for alloy in df['Alloy'].unique():
    print("\n" + "="*60)
    print(f"Analysis for Alloy: {alloy}")
    print("="*60)

    data = df[df['Alloy'] == alloy]

    # -----------------------------------
    # Visualization
    # -----------------------------------
    plt.figure(figsize=(8,6))
    sns.boxplot(data=data, x='Method', y='Hardness', hue='Dentist')
    plt.title(f'Boxplot: Dentist vs Method (Alloy {alloy})')
    plt.show()

    # Interaction Plot
    plt.figure(figsize=(8,6))
    sns.pointplot(data=data, x='Method', y='Hardness', hue='Dentist', markers='o')
    plt.title(f'Interaction Plot (Alloy {alloy})')
    plt.show()

    # -----------------------------------
    # Assumption Checks
    # -----------------------------------

    # Fit model
    model = ols('Hardness ~ C(Dentist) * C(Method)', data=data).fit()

    # Normality (Shapiro)
    shapiro_test = shapiro(model.resid)
    print("\nShapiro Test (Normality):", shapiro_test)

    # Homogeneity (Levene)
    groups = [group["Hardness"].values for name, group in data.groupby(['Dentist','Method'])]
    levene_test = levene(*groups)
    print("Levene Test (Equal Variance):", levene_test)

    # -----------------------------------
    # Two-Way ANOVA
    # -----------------------------------
    anova_table = sm.stats.anova_lm(model, typ=2)
    print("\nANOVA Table:\n", anova_table)

    # -----------------------------------
    # Post-hoc (Tukey Test)
    # -----------------------------------

    # The 'data' DataFrame already has 'Hardness' due to global rename.
    # No need to rename again here: data = data.rename(columns={"Response": "Hardness"})

    # Combine Dentist + Method
    data['Interaction'] = data['Dentist'].astype(str) + "-" + data['Method'].astype(str)

    tukey = pairwise_tukeyhsd(endog=data['Hardness'],
                             groups=data['Interaction'],
                             alpha=0.05)

    print("\nTukey HSD Results:\n")
    print(tukey)

    # Plot Tukey
    tukey.plot_simultaneous()
    plt.title(f'Tukey Test (Alloy {alloy})')
    plt.show()
```
